```python
import pandas as pd
import numpy as np
from scipy.stats import spearmanr
import csv
```

# Load Data

```python
file_path = "ergebnisse_nummeriert.csv"

with open(file_path, "r", encoding="utf-8-sig") as f:
    sample = f.read(5000)
    dialect = csv.Sniffer().sniff(sample)

df = pd.read_csv(
    file_path,
    sep=dialect.delimiter,
    encoding="utf-8-sig",
    engine="python",
    on_bad_lines="warn"
)

df.head()
```

# Variables

```python
fam_cols = {
    "ML": "Familiarity.Familiarity with Machine Learning",
    "CreditRisk": "Familiarity.Familiarity with Credit Risk Models",
    "Python": "Familiarity.Familiarity with Python"
}
```

```python
# all columns of the four contructs
interpret_cols = [c for c in df.columns if "Interpretability" in c]
reg_cols = [c for c in df.columns if "Regulatory Plausibility" in c]
trust_cols = [c for c in df.columns if "Trust" in c]
accept_cols = [c for c in df.columns if "Acceptance" in c]
```

```python
# mean per participant
df["Interpretability_mean"] = df[interpret_cols].mean(axis=1)
df["Regulatory_mean"] = df[reg_cols].mean(axis=1)
df["Trust_mean"] = df[trust_cols].mean(axis=1)
df["Acceptance_mean"] = df[accept_cols].mean(axis=1)

targets = [
    "Interpretability_mean",
    "Regulatory_mean",
    "Trust_mean",
    "Acceptance_mean"
]
```

# Spearman correlation

```python
for fam_name, fam_col in fam_cols.items():

    print(f"\n=== {fam_name} ===")

    for target in targets:

        rho, p = spearmanr(
            df[fam_col],
            df[target],
            nan_policy="omit"
        )

        print(
            f"{target:25s} "
            f"rho={rho:6.3f} "
            f"p={p:6.4f}"
        )
```
