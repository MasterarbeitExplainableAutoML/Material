```python
import pandas as pd

# Performance values
results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "AutoGluon L1",
        "AutoGluon L2",
        "AutoGluon L3",
        "AutoGluon L4",
        "H2O L1",
        "H2O L2",
        "H2O L3",
        "H2O L4"
    ],
    "Gini": [
        0.472608,
        0.513795,
        0.514009,
        0.513732,
        0.515136,
        0.499585,
        0.499585,
        0.499585,
        0.508930
    ],
    "Brier": [
        0.095659,
        0.098112,
        0.098412,
        0.098614,
        0.094243,
        0.114623,
        0.114623,
        0.114623,
        0.095990
    ]
})

# Baseline
gini_base = results.loc[0, "Gini"]
brier_base = results.loc[0, "Brier"]

# Percentage changes
results["Δ Gini (%)"] = (results["Gini"] - gini_base) / gini_base * 100
results["Δ Brier (%)"] = (results["Brier"] - brier_base) / brier_base * 100

# Logistic Regression has no delta
results.loc[0, ["Δ Gini (%)", "Δ Brier (%)"]] = None

# Format
results["Gini"] = results["Gini"].map(lambda x: f"{x:.4f}")
results["Brier"] = results["Brier"].map(lambda x: f"{x:.4f}")

results["Δ Gini (%)"] = results["Δ Gini (%)"].map(
    lambda x: "--" if pd.isna(x) else f"{x:+.2f}"
)

results["Δ Brier (%)"] = results["Δ Brier (%)"].map(
    lambda x: "--" if pd.isna(x) else f"{x:+.2f}"
)

print(results)
```
